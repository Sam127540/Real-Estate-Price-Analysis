```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df  = pd.read_csv("/content/kc_house_data_NaN.csv")

print("Dataset loaded successfully")
print("Dataset shape:, df.shape")

df.head

```

```python
print("Column names:")
print(df.columns.tolist())

print("Dataset Information:")
df.info()

print("Missing values in each column:")
print(df.isnull().sum())
```

```python
# Create a copy of the original dataset
df_clean = df.copy()

# 1. Remove the unnecessary index column
if 'Unnamed: 0' in df_clean.columns:
    df_clean = df_clean.drop(columns=['Unnamed: 0'])

# 2. Convert date into datetime format
df_clean['date'] = pd.to_datetime(
    df_clean['date'],
    errors='coerce'
)

# 3. Check duplicate rows before removing them
print("Duplicate rows before cleaning:",
      df_clean.duplicated().sum())
# 4. Remove rows with missing important values
df_clean = df_clean.dropna(
    subset=['bedrooms', 'bathrooms', 'date']
)

# 5. Remove duplicate rows
df_clean = df_clean.drop_duplicates()

# 6. Convert bedrooms into integer format
df_clean['bedrooms'] = df_clean['bedrooms'].astype(int)

# 7. Check the cleaned dataset
print("\nCleaned dataset shape:", df_clean.shape)

print("\nMissing values after cleaning:")
print(df_clean.isnull().sum())

print("\nUpdated data types:")
print(df_clean.dtypes)

# Display first 5 cleaned records
df_clean.head()

```

```python
# Basic Statistical Analysis

print("Total records:", len(df_clean))

print(f"Average house price: ${df_clean['price'].mean():,.2f}")

print(f"Median house price: ${df_clean['price'].median():,.2f}")

print(f"Minimum house price: ${df_clean['price'].min():,.2f}")

print(f"Maximum house price: ${df_clean['price'].max():,.2f}")

print("\nStatistical summary of house prices:")
print(df_clean['price'].describe())
```

```python
# House Price Distribution

plt.figure(figsize=(10, 5))

sns.histplot(
    data=df_clean,
    x='price',
    bins=50,
    color='skyblue'
)

plt.title('Distribution of House Prices', fontsize=16)
plt.xlabel('House Price (USD)')
plt.ylabel('Number of Houses')
plt.xlim(0, 2_000_000)
plt.tight_layout()
plt.show()
```

**Analysis**:

The histogram shows the distribution of house prices in the dataset. Most houses are concentrated in the lower price ranges, while a smaller number of houses have very high prices. This creates a right-skewed distribution.

```python
# Analyze houses with 1 to 6 bedrooms
bedroom_data = df_clean[
    df_clean['bedrooms'].between(1, 6)
]

# Calculate average price and number of houses
bedroom_analysis = (
    bedroom_data.groupby('bedrooms')
    .agg(
        average_price=('price', 'mean'),
        house_count=('price', 'count')
    )
    .reset_index()
)

print("Bedroom-wise analysis:")
print(bedroom_analysis.round(2))

# Create bar chart
plt.figure(figsize=(9, 5))

sns.barplot(
    data=bedroom_analysis,
    x='bedrooms',
    y='average_price',
    color='skyblue'
)

plt.title('Average House Price by Number of Bedrooms')
plt.xlabel('Number of Bedrooms')
plt.ylabel('Average House Price (USD)')
plt.ticklabel_format(style='plain', axis='y')
plt.tight_layout()
plt.show()
```

**Analysis**:

The bar chart compares the average house prices for houses with different numbers of bedrooms. The average price varies across bedroom groups. However, the price does not always increase with the number of bedrooms because other factors, such as living area, location, and property condition, also affect house prices.

```python
# Relationship between living area and house price

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df_clean,
    x='sqft_living',
    y='price',
    alpha=0.3,
    s=20
)

plt.title('House Price vs Living Area')
plt.xlabel('Living Area (Square Feet)')
plt.ylabel('House Price (USD)')
plt.ticklabel_format(style='plain', axis='y')
plt.tight_layout()
plt.show()
```

**Analysis**:

The scatter plot shows the relationship between house prices and living area. In general, houses with larger living areas tend to have higher prices. However, some houses with similar living areas have different prices, indicating that other factors, such as location, condition, and property grade, may also influence prices.

```python
# Calculate average price and number of houses by ZIP code

location_analysis = (
    df_clean.groupby('zipcode')
    .agg(
        average_price=('price', 'mean'),
        house_count=('price', 'count')
    )
    .reset_index()
)

# Select top 10 ZIP codes by average house price
top_locations = (
    location_analysis
    .sort_values('average_price', ascending=False)
    .head(10)
    .sort_values('average_price', ascending=True)
)

print("Top 10 ZIP codes by average house price:")
print(
    top_locations.sort_values(
        'average_price', ascending=False
    ).round(2).to_string(index=False)
)

# Create horizontal bar chart
plt.figure(figsize=(11, 6))

sns.barplot(
    data=top_locations,
    x='average_price',
    y='zipcode',
    color='skyblue'
)

plt.title('Top 10 ZIP Codes by Average House Price')
plt.xlabel('Average House Price (USD)')
plt.ylabel('ZIP Code')

# Removed ticklabel_format to avoid the AttributeError
plt.tight_layout()
plt.show()
```

**Analysis**:

The bar chart displays the ten zip codes with the highest average house prices in the dataset. The average prices vary across these locations, suggesting that location may play an important role in determining property values.

```python
# Select important numerical columns
columns = [
    'price',
    'bedrooms',
    'bathrooms',
    'sqft_living',
    'sqft_lot',
    'floors',
    'waterfront',
    'view',
    'condition',
    'grade',
    'sqft_above',
    'sqft_basement',
    'yr_built',
    'yr_renovated',
    'sqft_living15',
    'sqft_lot15'
]

# Calculate correlation
correlation = df_clean[columns].corr()

# Plot correlation heatmap
plt.figure(figsize=(13, 9))

sns.heatmap(
    correlation,
    cmap='coolwarm',
    annot=True,
    fmt='.2f',
    linewidths=0.5
)

plt.title('Correlation Heatmap of House Features')
plt.tight_layout()
plt.show()
```

**Analysis**:

The correlation heatmap shows the strength and direction of relationships between numerical features in the housing dataset. Positive values indicate that two features tend to increase together, while negative values indicate that one tends to decrease as the other increases. Values close to zero suggest a weak linear relationship.

```python
# Find which features have the strongest correlation with price

price_correlation = (
    df_clean.corr(numeric_only=True)['price']
    .drop('price')
    .sort_values(ascending=False)
)

print("Correlation of features with house price:")
print(price_correlation.round(3))
```

```python

# Extract year and month from the date column
df_clean['sale_month'] = (
    df_clean['date'].dt.to_period('M').astype(str)
)

# Calculate monthly average price and number of sales
monthly_analysis = (
    df_clean.groupby('sale_month')
    .agg(
        average_price=('price', 'mean'),
        houses_sold=('id', 'count')
    )
    .reset_index()
)

print("Monthly House Price Analysis:")
print(monthly_analysis.round(2).to_string(index=False))

# Plot monthly average house prices
plt.figure(figsize=(12, 5))

sns.lineplot(
    data=monthly_analysis,
    x='sale_month',
    y='average_price',
    marker='o'
)

plt.title('Monthly Average House Price Trend')
plt.xlabel('Month')
plt.ylabel('Average House Price (USD)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

**Analysis**:

The line chart shows how the average house price changes over different months in the dataset. The trend helps us observe monthly variations in average prices and identify periods with relatively higher or lower average prices.
